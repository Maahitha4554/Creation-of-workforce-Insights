```python
#2. Create the Employee table
```

```python
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees (
    Employee_ID INTEGER PRIMARY KEY,
    Employee_Name TEXT,
    Age INTEGER,
    Gender TEXT,
    Department TEXT,
    Job_Role TEXT,
    Location TEXT,
    Education TEXT,
    Experience_Years INTEGER,
    Salary REAL,
    Performance_Score REAL,
    Job_Satisfaction INTEGER,
    Engagement_Score INTEGER,
    Training_Hours INTEGER,
    Promotions INTEGER,
    Remote_Work TEXT,
    Attrition TEXT,
    Joining_Date DATE
)
""")

conn.commit()

print("Employees table created successfully!")
```

```python
#4. Display all employees
```

```python
query = """
SELECT * FROM Employees;
"""

df = pd.read_sql_query(query, conn)
df
```

```python
#Query 1 — Count total employees
query = """
SELECT COUNT(*) AS Total_Employees
FROM Employees;
"""

pd.read_sql_query(query, conn)
```

```python
#2 — Employees by department
query = """
SELECT Department, COUNT(*) AS Employee_Count
FROM Employees
GROUP BY Department;
"""

pd.read_sql_query(query, conn)
```

```python
#3 — Average salary by department
query = """
SELECT
    Department,
    ROUND(AVG(Salary), 2) AS Average_Salary
FROM Employees
GROUP BY Department;
"""

pd.read_sql_query(query, conn)
```

```python
#4 — Average performance by department
query = """
SELECT
    Department,
    ROUND(AVG(Performance_Score), 2) AS Average_Performance
FROM Employees
GROUP BY Department;
"""

pd.read_sql_query(query, conn)
```

```python
#5 — Find employees with high performance
query = """
SELECT Employee_Name, Department, Job_Role, Performance_Score
FROM Employees
WHERE Performance_Score >= 4.5;
"""

pd.read_sql_query(query, conn)
```

```python
#Attrition Analysis
#6 — Employees who left
query = """
SELECT *
FROM Employees
WHERE Attrition = 'Yes';
"""

pd.read_sql_query(query, conn)
```

```python
#7 — Attrition count
query = """
SELECT
    Attrition,
    COUNT(*) AS Employee_Count
FROM Employees
GROUP BY Attrition;
"""

pd.read_sql_query(query, conn)
```

```python
#8 — Attrition rate
query = """
SELECT
    ROUND(
        100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*), 2
    ) AS Attrition_Rate
FROM Employees;
"""

pd.read_sql_query(query, conn)
```

```python
#9 — Attrition by department
query = """
SELECT
    Department,
    COUNT(*) AS Total_Employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS Attrition_Count,
    ROUND(
        100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) / COUNT(*), 2
    ) AS Attrition_Rate
FROM Employees
GROUP BY Department;
"""

pd.read_sql_query(query, conn)
```
